```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
h1 = pd.read_csv("results/astar_results/astar_h1_generated.csv")
h2 = pd.read_csv("results/astar_results/astar_h2_generated.csv")
h3 = pd.read_csv("results/astar_results/astar_h3_generated.csv")
h1_gt = pd.read_csv("results/astar_results/astar_gt_h1_generated.csv")
h2_gt = pd.read_csv("results/astar_results/astar_gt_h2_generated.csv")
h3_gt = pd.read_csv("results/astar_results/astar_gt_h3_generated.csv")
```

```python
medium_expanded = pd.DataFrame({
    "h1": h1[h1.run_id.str.contains("medium")].expanded,
    "h2": h2[h2.run_id.str.contains("medium")].expanded,
    "h3": h3[h3.run_id.str.contains("medium")].expanded,
    "h1_gt": h1_gt[h1_gt.run_id.str.contains("medium")].expanded,
    "h2_gt": h2_gt[h2_gt.run_id.str.contains("medium")].expanded,
    "h3_gt": h3_gt[h3_gt.run_id.str.contains("medium")].expanded,
})
medium_frontier = pd.DataFrame({
    "h1": h1[h1.run_id.str.contains("medium")].max_frontier,
    "h2": h2[h2.run_id.str.contains("medium")].max_frontier,
    "h3": h3[h3.run_id.str.contains("medium")].max_frontier,
    "h1_gt": h1_gt[h1_gt.run_id.str.contains("medium")].max_frontier,
    "h2_gt": h2_gt[h2_gt.run_id.str.contains("medium")].max_frontier,
    "h3_gt": h3_gt[h3_gt.run_id.str.contains("medium")].max_frontier,
})
```

```python
ax = sns.stripplot(data=medium_frontier, jitter=0.2, size=6, palette="Set2", alpha=0.8)
ax.set_title("Maximum Frontier")
ax.set_xlabel("Heuristic")
ax.set_ylabel("Nodes")
plt.savefig(
    'results/astar_results/figures/medium_frontier.png', # Nombre del archivo y extensión (.png, .jpg, .pdf)
    dpi=300,                       # Calidad de la imagen (300 DPI es excelente para reportes)
    bbox_inches='tight',           # Ajusta los márgenes automáticamente para que no se corte el título o los ejes
    transparent=False              # True si quieres el fondo transparente
)
```

```python
ax = sns.stripplot(data=medium_expanded, jitter=0.2, size=6, palette="Set2", alpha=0.8)
ax.set_title("Expanded Nodes")
ax.set_xlabel("Heuristic")
ax.set_ylabel("Nodes")
plt.savefig(
    'results/astar_results/figures/medium_expanded.png', # Nombre del archivo y extensión (.png, .jpg, .pdf)
    dpi=300,                       # Calidad de la imagen (300 DPI es excelente para reportes)
    bbox_inches='tight',           # Ajusta los márgenes automáticamente para que no se corte el título o los ejes
    transparent=False              # True si quieres el fondo transparente
)
```
